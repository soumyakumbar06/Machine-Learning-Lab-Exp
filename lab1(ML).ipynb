```python
import math
print(math.sqrt(25))
print(math.factorial(5))
print(round(math.pi,2))
print(round(math.e,2))
print(round(math.sin(math.radians(30)),2))
```

```python
import matplotlib.pyplot as plt
labels =['A','B','C','D']
sizes=[10,20,15,25]
plt.pie(sizes,labels=labels,autopct='%1.1f%%',colors=['blue','red','green','yellow'])
plt.title('Pie Chart Example')
plt.show()
```

```python
categories=['A','B','C','D']
values=[10,20,15,25]
plt.bar(categories,values,color='skyblue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart Example')
plt.show()
```

```python
import numpy as np
data=np.random.randn(1000)
plt.hist(data,bins=20,color='yellow',edgecolor='blue')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram Example')
plt.show()
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y,color='pink',marker='x')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter Plot Example')
plt.show()
```

```python
import pandas as pd
# Since '/data5.csv' is not present, we create the DataFrame from the dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 700000]
}
df = pd.DataFrame(data)
print(df.head())
print(df.head(3))
print(df.tail(2))
print(df.shape)
print(df.info())
print(df.describe())
print(df[df['Age']>28])
print(df[(df['Age']>25)])
```
