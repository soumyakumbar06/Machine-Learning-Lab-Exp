```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error

# Fixed the path from absolute '/...' to relative '...'
df=pd.read_csv("linear_regression_lab_15_records.csv")
print(df)

x=df[["Hours_Studied"]]
y=df["Marks"]

print("\n\nX: ",x)
print("\n\nY:\n",y)

kf = KFold(n_splits=5,shuffle=True)
print("Number of folds:", kf.n_splits)
mse_values=[]
fold=1

for train_index, validation_index in kf.split(x):

  x_train=x.iloc[train_index]
  y_train=y.iloc[train_index]
  x_validation=x.iloc[validation_index]
  y_validation=y.iloc[validation_index]

  model=LinearRegression()
  model.fit(x_train,y_train)

  slope=model.coef_[0]
  intercept=model.intercept_
  y_predict=model.predict(x_validation)

  mse=mean_squared_error(y_validation,y_predict)
  mse_values.append(mse)
  print("\n\n==========FOLD",fold,"==========")
  print("Training records:",list(train_index))
  print("Validation records:",list(validation_index))
  print("Slope:",round(slope,4))
  print("Intercept:",round(intercept,4))
  print("Regression Equation:")
  print("Y =",round(intercept,4),"+",round(slope,4),"X")
  print("Actual values:",list(y_validation))
  print("Predicted values:",[round(value,4)for value in y_predict])
  print("Validation MSE:",round(mse,4))
  print()
  fold+=1
  print("MSE for each fold: ")
  for i, mse in enumerate(mse_values, start=1):
    print("Fold ",i,"MSE =",round(mse,4))
    avrage_mse=sum(mse_values)/len(mse_values)
    print("Average Cross-Validation MSE: ",round(avrage_mse,4))
```
